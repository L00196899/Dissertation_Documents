```python
# Test JSON Payload

payload = {
    "EmployeeInvoicePayload": {
        "employee_id": "128213",
        "full_name": "Sachin Khedkar",
        "email": "sachin_k@gmail.com",
        "phone_number": "+919876543210",
        "joining_date": "2025-05-10",
        "country": "India",
        "city": "Bangalore",
        "timezone": "Asia/Kolkata",
        "designation": "Senior Data Scientist",
        "department": "AI & Analytics",
        "business_unit": "Finance",
        "project_description": "Building invoice automation workflow for finance operations",
        "priority": "HIGH",
        "tags": [
            "finance",
            "invoice",
            "automation"
        ],
        "manager_name": "Amit Verma",
        "salary_currency": "INR",
        "work_mode": "Hybrid",
        "access_level": "Level_2",
        "retry_count": 2,
        "payload_source": "Postman",
        "transaction_type": "Vendor Invoice Processing",
        "risk_level": "MEDIUM",
        "remarks": "Intermittent payment processing timeout observed during peak hours"
    }
}
```

```python
headers = {
    "Content-Type": "application/json"
}
```

```python
# API Call

try:

    print("\n SENDING PAYLOAD : \n")
    print(json.dumps(payload, indent=4))

    response = requests.post(
        SAP_CPI_URL,
        headers=headers,
        auth=(USERNAME, PASSWORD),
        json=payload,
        timeout=30
    )

    print("\n RESPONSE RECEIVED :")

    print(f"\nStatus Code: {response.status_code}")

    print("\nResponse Headers:")
    print(response.headers)

    try:
        response_json = response.json()

        print("\nResponse JSON:")
        print(json.dumps(response_json, indent=4))

    except:
        print("\nResponse Text:")
        print(response.text)

    # Response check

    if response.status_code in [200, 201, 202]:
        print("\n SUCCESS: SAP CPI received the payload successfully.")

    else:
        print("\n FAILURE: SAP CPI rejected the payload.")


# Error handling

except requests.exceptions.Timeout:
    print("\n ERROR: Request timed out.")

except requests.exceptions.ConnectionError:
    print("\n ERROR: Unable to connect to SAP CPI endpoint.")

except requests.exceptions.RequestException as e:
    print(f"\n REQUEST ERROR: {e}")

except Exception as e:
    print(f"\n UNEXPECTED ERROR: {e}")
```
